```python
# Test case 2: Zero Matching Rows
# Test to see if it will filter any rows above the condition

normal_test = df[df['wage-increase-first-year'] > 7.0]
print(normal_test)
print('\nAmount of wage increases above 7%: ', normal_test['wage-increase-first-year'].count())
```

```python
#Test Case 3: Exactly one matching row
# Checks and sees which/how many rows increase exactly by 7%

normal_test = df[df['wage-increase-first-year'] == 7.0]
print(normal_test.shape[0])
print('\nAmount of wage increases exactly at 7%: ', normal_test['wage-increase-first-year'].count())
```

```python
# Test Case 4: Multiple Matching Rows
#Checks how many rows have a wage increase above 5%

multiple_test = df[df['wage-increase-first-year'] > 5.0]

print(multiple_test)
print("\nNumber of matching rows:", multiple_test.shape[0])
```

```python
# Test Case 5: Missing Values
# Checks how many values are reported as NaN

missing_test = df['wage-increase-first-year']

print("Missing values:", missing_test.isna().sum())
print("Non-missing values:", missing_test.count())
print("Average wage increase:", missing_test.mean())
```
